```python
# count the number of digits in a number
num = int(input("Enter a number: "))

count = 0

if num == 0:
    count = 1
else:
    while num > 0:
        num = num // 10
        count += 1

print("Number of digits:", count)
#Step by step:
#Start: num = 12345, count = 0
#12345 > 0 → num = 1234, count = 1
#1234 > 0 → num = 123, count = 2
#123 > 0 → num = 12, count = 3
#12 > 0 → num = 1, count = 4
#1 > 0 → num = 0, count = 5
#Now num is 0, so the loop stops
#Final output:


#Why this works
#num = num // 10 removes the last digit each time, so the loop keeps reducing the number until it becomes 0. The count variable stores how many times this happened.
```

```python
# sum digits in a number
num = int(input("Enter a number: "))
sum_digit = 0

while num > 0:
    digit = num % 10
    sum_digit += digit
    num //= 10

print("Sum of digits:", sum_digit)
#dry run
#(num = 1234, sum_digit = 0
#digit = 1234 % 10 = 4 → sum_digit = 4 → num = 123
#digit = 123 % 10 = 3 → sum_digit = 7 → num = 12
#digit = 12 % 10 = 2 → sum_digit = 9 → num = 1
#digit = 1 % 10 = 1 → sum_digit = 10 → num = 0
#Loop stops because num becomes 0.

#Final output:


#If you want, I can also give you the same answer in a very short “school notebook” format.)
```

```python
# palindrome program
num = int(input("Enter a number: "))
original = num
reverse = 0

while num > 0:
    digit = num % 10
    reverse = reverse * 10 + digit
    num //= 10

if original == reverse:
    print("Palindrome")
else:
    print("Not a palindrome")
```

```python
# check if the number is prime
number = int(input("Enter a number: "))
count = 0

for i in range(1, number + 1):
    if number % i == 0:
        count += 1

if count == 2:
    print("Prime number")
else:
    print("Not a prime number")
```
